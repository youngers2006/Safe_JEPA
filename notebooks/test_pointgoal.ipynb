imports

```python
import numpy as np
import safety_gymnasium as sgym
import gymnasium as gym
import matplotlib.pyplot as plt
```

env setup

```python
env_name = "SafetyPointGoal1-v0"
env = sgym.make(env_name, render_mode="human")
```

Run env

```python
obs, _ = env.reset(seed=10)
done = 0

t = 0
rewards = []
costs = []
while not done:
    action = env.action_space.sample()
    next_obs, reward, cost, terminated, truncated, _ = env.step(action)
    done = (terminated or truncated)
    if t == 1000:
        done = True
    t += 1
    rewards.append(reward)
    costs.append(cost)
```

Plotting

```python
plt.plot(rewards)
```

```python
plt.plot(costs)
print(costs)
```
